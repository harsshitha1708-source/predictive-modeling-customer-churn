```python
# ============================================================
# PREDICTIVE MODELING USING MACHINE LEARNING
# CUSTOMER CHURN PREDICTION
# ============================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_curve,
    roc_auc_score
)

print("Libraries imported successfully!")


# -----------------------------
# 2. UPLOAD DATASET
# -----------------------------

uploaded = files.upload()

# Read the uploaded CSV file
df = pd.read_csv("customer-churn.csv.csv")

print("\nDataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])


# -----------------------------
# 3. DISPLAY DATASET
# -----------------------------

print("\nFirst 5 rows:")
display(df.head())

print("\nDataset information:")
df.info()


# -----------------------------
# 4. CHECK MISSING VALUES
# -----------------------------

print("\nMissing values before cleaning:")
print(df.isnull().sum())


# -----------------------------
# 5. CLEAN TOTALCHARGES
# -----------------------------

# Convert TotalCharges from text to numeric
# Invalid/blank values become NaN

df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

# Fill missing TotalCharges using median
df["TotalCharges"] = df["TotalCharges"].fillna(
    df["TotalCharges"].median()
)

print("\nMissing values after cleaning:")
print(df.isnull().sum())


# -----------------------------
# 6. CHECK DUPLICATES
# -----------------------------

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())


# -----------------------------
# 7. REMOVE CUSTOMER ID
# -----------------------------

# Customer ID is only an identifier.
# It does not provide useful predictive information.

df = df.drop("customerID", axis=1)


# -----------------------------
# 8. CONVERT TARGET VARIABLE
# -----------------------------

# Convert:
# Yes -> 1
# No  -> 0

df["Churn"] = df["Churn"].map({
    "Yes": 1,
    "No": 0
})


# -----------------------------
# 9. ENCODE CATEGORICAL VARIABLES
# -----------------------------

# Convert categorical columns into numerical columns

categorical_columns = df.select_dtypes(
    include=["object"]
).columns

df = pd.get_dummies(
    df,
    columns=categorical_columns,
    drop_first=True
)

# Convert True/False values into 1/0
df = df.astype(int)


print("\nData after preprocessing:")
display(df.head())


# -----------------------------
# 10. SEPARATE FEATURES AND TARGET
# -----------------------------

X = df.drop("Churn", axis=1)

y = df["Churn"]

print("\nNumber of input features:", X.shape[1])
print("Target variable: Churn")


# -----------------------------
# 11. SPLIT DATASET
# -----------------------------

# 80% training
# 20% testing

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])


# -----------------------------
# 12. CREATE RANDOM FOREST MODEL
# -----------------------------

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)


# -----------------------------
# 13. TRAIN MODEL
# -----------------------------

model.fit(X_train, y_train)

print("\nRandom Forest model trained successfully!")


# -----------------------------
# 14. MAKE PREDICTIONS
# -----------------------------

y_pred = model.predict(X_test)

print("Predictions generated successfully!")


# -----------------------------
# 15. CALCULATE ACCURACY
# -----------------------------

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\n==============================")
print("MODEL ACCURACY")
print("==============================")

print("Accuracy:", accuracy)
print("Accuracy Percentage:", round(accuracy * 100, 2), "%")


# -----------------------------
# 16. CLASSIFICATION REPORT
# -----------------------------

print("\n==============================")
print("CLASSIFICATION REPORT")
print("==============================")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=["Stayed", "Churned"]
    )
)


# -----------------------------
# 17. CONFUSION MATRIX
# -----------------------------

cm = confusion_matrix(
    y_test,
    y_pred
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Stayed", "Churned"],
    yticklabels=["Stayed", "Churned"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix - Customer Churn Prediction")

plt.show()


# -----------------------------
# 18. ROC CURVE
# -----------------------------

# Get probability of churn
y_probability = model.predict_proba(X_test)[:, 1]

# Calculate ROC values
fpr, tpr, thresholds = roc_curve(
    y_test,
    y_probability
)

# Calculate AUC
auc_score = roc_auc_score(
    y_test,
    y_probability
)

plt.figure(figsize=(7, 5))

plt.plot(
    fpr,
    tpr,
    label=f"Random Forest (AUC = {auc_score:.2f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Customer Churn Prediction")

plt.legend()

plt.show()

print("\nROC-AUC Score:", round(auc_score, 4))


# -----------------------------
# 19. FEATURE IMPORTANCE
# -----------------------------

feature_importance = pd.Series(
    model.feature_importances_,
    index=X.columns
)

feature_importance = feature_importance.sort_values(
    ascending=False
).head(10)

plt.figure(figsize=(8, 5))

feature_importance.sort_values().plot(
    kind="barh"
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Top 10 Important Features")

plt.show()


# -----------------------------
# 20. FINAL RESULT
# -----------------------------

print("\n======================================")
print("FINAL PROJECT RESULTS")
print("======================================")

print("Algorithm: Random Forest Classifier")
print("Training Samples:", X_train.shape[0])
print("Testing Samples:", X_test.shape[0])
print("Accuracy:", round(accuracy * 100, 2), "%")
print("ROC-AUC:", round(auc_score, 4))

print("\nProject completed successfully!")
```
